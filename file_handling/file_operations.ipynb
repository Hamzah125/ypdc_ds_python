# file operations/modes in file operations 
read a file : r
write a file: w
add to a file: a
write to a binary file: wb
read a binary file: rb
read,write,truncate: w+



```python
with open('description.txt', 'r') as f:
    content = f.read()

print(content)
```

```python
with open('example.txt', 'r') as file:
    context = file.read()
    
print(context)
    
```

```python
# overriding a file

with open('example.txt', 'w') as file:
    file.write('I am hamza\n')
    file.write('i am peer mentor\n')
    file.write('i try to guide my fellow\n')
    
print("file has been edited")
```

```python
# adding something to the existing

with open('example.txt', 'a') as file:
    file.write('i am a reader\n')
    file.write('rn i am reading russian literature')
    
with open('example.txt', 'r') as file:
    print(file.read())
```

```python
# iteration

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

Binary files are machine readable files , e.g .exe, png, jpg


```python
# writing a binary file

data = b'\x46\x65\x03\x67\x61\x74\x61'

with open('binary_file.bin', 'wb') as binary_file:
    binary_file.write(data)
    
print("Binary file created successfully.")
    

```

```python
# to read a binary file

with open('binary_file.bin', 'rb') as b:
    sage = b.read()
    
print(sage)
```

```python
# writing and reading at the same time

with open('my_new.txt', 'w+') as file:
    file.write("Hello, World!\n")
    file.write("This is a test file.\n")
    file.write("Let's read it back.")
    file.seek(0)
    content = file.read()
    print(content)

# writing and reading at the same time with binary file
with open('my_new_binary.bin', 'w+b') as binary_file:
    binary_file.write(b'\x46\x65\x03\x67\x61\x74\x61')
    binary_file.seek(0)
    binary_content = binary_file.read()
    print(binary_content)
```
