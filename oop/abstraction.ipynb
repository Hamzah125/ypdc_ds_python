```python
# interface 

from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    @abstractmethod
    def pay(self, amount):
        pass
```

```python
class JazzCash(PaymentGateway):
    def pay(self, amount):
        return f"paid {amount} by JC"
```

```python
class GooglePay(PaymentGateway):
    def pay(self, amount):
        return f"paid {amount} by GP"
```

```python
for p in [JazzCash(), GooglePay()]:
    print(p.pay(100000))
```
