Polymorphism: same method name; deifferent behavior depending on objects.reduces the complexity, reusability, security and code modulation , it tells what not How

Polymorphism is same function/method which behaves differently depending on different object or datatype
it makes  our code more flexible, reusable, elegant, instead of writing multiple functions you can write one and make it interfere in diffrent situations

```python
# basic example

class Cow:
    def sound(self):
        return "moo"
```

```python
class Goat:
    def sound(self):
        return "baa"
```

```python
class Cat:
    def sound(self):
        return "meow"
```

```python
def animal_sound(animal):
    print(animal.sound())
    
c = Cow()
g = Goat()
c = Cat()
```

```python
animal_sound(c)
animal_sound(g)
animal_sound(c)
```

```python
# method overriding
# child class overrides parent class method

class Animal:
    def sound(self):
        return "animals make some sound"
```

```python
class Lion(Animal):
    def sound(self):
        return "roar"
    
class Wolf(Animal):
    def sound(self):
        return "howls"
```

```python
animals = [Animal(), Lion(), Wolf()]

for vertebrates in animals:
    print(vertebrates.sound())
```

```python
# abstraction in polymorphism

# we force subclasses to implement methods differently

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
```

```python
class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r
```

```python
class Rectangle(Shape):
    def __init__(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        return self.l * self.b
```

```python
shape = Rectangle(5, 10)
shape2 = Circle(7)

print(shape.area())
print(shape2.area())
```

```python
# bioinformatics

from abc import ABC, abstractmethod
    
class DataProcessor(ABC):
    @abstractmethod
    def procedure(self, data):
         pass
            
```

```python
class DNAprocessor(DataProcessor):
    def procedure(self, data):
        return data.replace("T", "U")
```

```python
class ProteinProcessor(DataProcessor):
    def procedure(self,data):
        return f"Amino Acids: {data}"
```

```python
processors = [DNAprocessor(), ProteinProcessor()]

seq = "ATGCTGCACGTA"
```

```python
for p in processors:
    print(p.procedure(seq))
```
