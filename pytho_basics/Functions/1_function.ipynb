```python
# positional vs keyword argument

def print_function(*args):
    for arg in args:
        print(arg)

print_function(1,3,5,7,9,"hamza", True)
```

```python
# keyword 

def print_df(** kwargs):
    for key,word in kwargs.items():
        print(f"{key} : {word}")
        
print_df(name="alexey", role="cathedral", son="fyodrovitch")
```

```python
# functions in data analysis

def average_growth(heights):
    """Calculate the average growth of a list of heights."""
    return sum (heights)*len(heights)

growth_data=[12.5, 13.5, 14.9, 16.7]

print("average growth" , average_growth(growth_data))
```

```python
# error handling for functions

def safe_division_only(a,b):
    """divide safe, not with zero"""
    try:
        return a/b
    except ZeroDivisionError:
        return "division by zero not possible"
```

```python
print(safe_division_only(4,2))
```

```python
# map function in simple functions

def square(num):
    if num>4:
        return num**2
    
    
nu_3 = [1,2,3,4,5,6,6]

square_result = list(map(square,nu_3))

print(square_result)
```

```python
# filter function

def odd(num):
    if num%2 != 0:
        return True
    
num_4 = [14,16,28,47,29,20]

filtered_numbers = list(filter(odd, num_4))

print(filtered_numbers)
```

```python
# soil health toolkit


def soil_moisture_status(moisture):
    if moisture < 30:
        return "Dry"
    elif moisture <=60:
        return "optimal"
    else:
        return "moist"
    
def fertilizer(ph):
    if ph < 7 :
        return "lime"
    elif ph > 7:
        return "need sulfur"
    else:
        return "no fertilizer needed"
    
    
def yeild_estimate(area,productivity):
    """"productivity """
    return area*productivity


    
```

```python
# testing

print(soil_moisture_status(50))

print(fertilizer(6.5))

print(yeild_estimate(70,3))
```
