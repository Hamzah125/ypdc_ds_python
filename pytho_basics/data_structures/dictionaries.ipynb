```python
# curly braces 

intro = {"name":"hamza", "marks":54, "remarks": True}

# dict method

intro = dict(name = "hamza", marks = 54, remarks = True)

print(type(intro))
```

```python
# accesion

print(intro)

#using key inside square bracket

print(intro["name"])
print(intro["remarks"])

# get method

print(intro.get("marks"))
print(intro.get("remarks"))
print(intro.get("roll no"))
```

```python
# updating

# adding new key value

intro["age"]=21
print(intro)

# UPDATING VALUE OF A KEY

intro["Remarks"]=False
print(intro)
intro["marks"]=32
print(intro)
```

```python
#access key

keys = intro.keys()
print(keys)

# value

value = intro.values()
print(value)

#items

products = intro.items()
print(products)
```

```python
# removes and return the value key

intro.pop('remarks')
```

```python
# popitem last key value pair 

intro.popitem()
```

```python
my_data = {"age":21, "hobby":"reading", "position":"pmds"}
```

```python
# shallow copy

new_data = my_data.copy
print(new_data)

my_data["age"] = 22

print(new_data)
print(my_data)

clone = my_data.copy()

print(clone)
```

```python
for k in my_data:
    print("age")


# access a value 

for v in my_data:
    print("hobby", my_data["hobby"])
    
for value in my_data.values():
    print(value)
    
```

```python
for key,value in my_data.items():
    print(key,value)
```

```python
# real world example

sentence = "python is powerful and it is handy to use programming specifically python"
words = sentence.split()
word_count={}

for word in words:
    word_count[word]=word_count.get(word,0)+1
    
print(word_count)
```
