```python
students = pd.DataFrame({
    "id" : [1, 2, 3, 4],
   "name" : ["katya", "grushenka", "grigory", "ivan"]
})
```

```python
students
```

```python
marks = pd.DataFrame({
    "id" : [2, 3, 4, 5],
    "scores" : [45, 78, 89, 90]
})
```

```python
marks
```

```python
# inner join # attributes that exist in both

pd.merge(students, marks, on="id", how="inner")
```

```python
# outer join # like a union 

pd.merge(students, marks, on="id", how="outer")
```

```python
# left join # all fro left side # matching from right

pd.merge(students, marks, on="id", how="left")
```

```python
# right join # all from right matching fromleft

pd.merge(students, marks, on="id", how="right")
```

```python
# concatenate

pd.concat([students, marks], axis=0)
```
