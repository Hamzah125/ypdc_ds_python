sets
A set is UNORDERED COLLECTION CONTAINING UNIQUE ELEMENTS
_NO DUPLCATES _MUTABILITY IF ITEMS ARE _UNORDERED _MATHEMATICAL OPERATIONS

```python
# SYNTAX
my_set = {}
print(my_set)

your_set = {2,4,6,8,10}
print(your_set)
```

```python
# real world examples

crops = ["wheat", "maize", "rice", "cotton", "maize", "cotton"]
print(crops)

uniqueness = set(crops)
print(uniqueness)
```

```python
# biology

bases = ["A","G","C","A","G","A"]
unique_bases = set(bases)
print(unique_bases)
```

```python
# operators

set_1 = {1,3,5,7,9}

set_2 = {1,2,3,4,5}

# union

print(set_1|set_2)

# intersection

print(set_1&set_2)

# difference

print(set_1-set_2)

# symmetric difference

print(set_1^set_2)

```
