```python
# input_functionality of python

num_1 = input("your first number")
num_2 = input("your second number")
```

```python
print(num_1 , num_2)
```

```python
print(type(num_1))
print(type(num_2))
```

```python
type_casting_num_1 = int(num_1)
type_casting_num_2 = int(num_2)
```

```python
print(type(num_1))
print(type(type_casting_num_1))
```

```python
sum = (type_casting_num_1 + type_casting_num_2)
difference = (type_casting_num_1 - type_casting_num_2)
multiplication = (type_casting_num_1 * type_casting_num_2)
division = (type_casting_num_1 / type_casting_num_2)
```

```python
print(sum)
print(difference)
print(multiplication)
print(division)
```
