magic methods are special methods which are sorrounded by double underscores (__methods__)
python calls  them automatically when certain actions are performed
__init__ , object is created, 
__str__ , object is printed
__add__, + operartion
__repr__, object representation

```python
# Magic Methods in Python

class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
    def __str__ (self):
        return f"Student Name: {self.name}, Grade: {self.grade}"
```

```python
s1 = Student("ali", "c")
print(s1)
```

```python
# another example

class DNA:
    def __init__(self, sequence):
        self.sequence = sequence
    
    def __len__(self):
        return(len(self.sequence))
    
    def __getitem__(self, index):
        return self.sequence[index]
    
    def __str__(self):
        return self.sequence
    
    def __add__(self, other):
        return DNA(self.sequence + other.sequence)
```

```python
#usage
dna1 = DNA("ATCG")
dna2 = DNA("GGTA")
print(len(dna1))          
print(len(dna2))
print(dna1[1])
print(dna1+dna2)
```
