```python
# univariable analysis # numerical

data = { 
        "salary": [33000, 44000, 35000, 44000, 78000, 180000, 21000, 50000, 65000]}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.shape
```

```python
# line plot

plt.plot(df["salary"], marker = "*", color = "red",linestyle = "--", linewidth = 2)
plt.grid()
plt.title("Salary Distribution")
plt.xlabel("Index")
plt.ylabel("salary")
```

```python
# histogram  # to see the distribution of variable

plt.hist(df["salary"], bins= 10, color = "green", alpha = 0.6)
```

```python
# boxplot

plt.boxplot(df["salary"])
```

```python
df.info()
```

```python
# cetagorical

df["dept"] = ["IT", "FINANCE", "SALES", "IT", "HR", "OPERATIONS", "HR", "IT", "SALES"]
```

```python
df.head()
```

```python
# line chart and boxplot only works on numerical data

# pie chart # composition

count = df["dept"].value_counts()
count
```

```python
# pie plot
plt.pie(count, labels = count.index, autopct = "%1.1f", explode = (0, 0.1, 0.2, 0.1, 0.1))
plt.show()
```

```python
# countplot

plt.bar(count.index, count, color = ["yellow", "blue", "green", "orange", "red"])
plt.show()
```

```python
# bavariant # using two variables

# numerical- numerical

# suppose a variable-age

df ["age"] = [43, 45, 51, 57, 54, 50, 21, 44, 38]
```

```python
df.head()
```

```python
# scatter plot 

plt.scatter(df["age"], df["salary"], color = "red", marker = '*')
plt.show()
```

```python
sort_age = df["age"].sort_values()
sort_age
```

```python
# line plotting 

plt.plot(sort_age, df["salary"], marker = "o", linewidth = "2")
plt.grid()
plt.show()
```

```python
# bar chart

plt.bar(df["age"], df["salary"], color = "blue", width = 1)
plt.show()
```

```python
df
```

```python
# numerical and cetagorical

hr_sal = df[df["dept"] == "HR"]["salary"]
```

```python
hr_sal
```

```python
it_sal = df[df["dept"] == "IT"]["salary"]
it_sal
```

```python
sales_sal = df[df["dept"] == "SALES"]["salary"]
```

```python
sales_sal
```

```python
fin_sal = df[df["dept"] == "FINANCE"]["salary"]
fin_sal
```

```python
plt.boxplot([fin_sal, it_sal, sales_sal], labels = ["FINANCE", "IT", "SALES"])
plt.show()
            
```

```python
# numerical _ cetagorical

salary_by_dept = df.groupby("dept")["salary"].sum()
```

```python
salary_by_dept
```

```python
plt.pie(salary_by_dept, labels = salary_by_dept.index, autopct ="%1.1f")
plt.show()
```

```python
it_mean = sum(it_sal) / len(it_sal)
hr_mean = sum(hr_sal) / len(hr_sal)
fin_mean = sum(fin_sal) / len(fin_sal)
```

```python
it_mean
hr_mean
fin_mean
```

```python
plt.bar(["IT", "HR", "FINANCE"], [it_mean, hr_mean, fin_mean], color = ["blue", "orange", "green"])
plt.show()
```

```python
# multivariate analysis

# bubble plot # just a scatter plot where 3rd numerical column is defined by size of bubble 

df["experience"] = [1.1, 5.6,2.2, 3, 0.0, 5.6, 4.6, 2.5, 2.0]
```

```python
df.head()
```

```python
plt.scatter(df["age"], df["salary"], s= df["experience"]*20)

plt.show
```

```python
plt.scatter(df["age"], df["salary"], c = df["dept"].map({"HR": "yellow", "IT":"blue", "FINANCE": "orange"}))
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary vs Dept")
plt.legend()
plt.show()
```

```python
plt.scatter(df["age"], df["salary"],  c=  df["dept"].map({"IT" :"blue", "HR" :"orange", "FINANCE" :"green", "OPERATIONS" :"red", "SALES" :"purple"}))
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("age vs salary wrt dept")
plt.show()
```

```python
data2 = {
    "year": [2020, 2021, 2022, 2023],
    "sales": [150, 200, 250, 300],
    "expense": [100, 50, 65, 70],
    "profit": [50, 150, 185, 230]
}
```

```python
df2 = pd.DataFrame(data2)
```

```python
df2.head()
```

```python
plt.plot(df2["year"], df2["profit"], label = "profit", marker = '*')
plt.plot(df2["year"], df2["sales"], label = "sales", marker = "o")
plt.plot(df2["year"], df2["expense"], label = "expense",marker = "s" )
plt.xlabel("Year")
plt.ylabel("total")
plt.title("Company Performance")
plt.legend()
plt.show()
```

```python
# 3d 

axs = plt.axes(projection = "3d")

axs.scatter(df["age"], df["salary"], df["experience"])
axs.set_xlabel("Age")
axs.set_ylabel("salary")
axs.set_zlabel("Experience")
axs.set_title("3D Scatter plot")
plt.show()
```
