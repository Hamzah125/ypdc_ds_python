Tuples is a type of data structure which is ordered, immutable and iterable

Immutability helps us in:
integrity
hashability
performance

```python
#syntax
nyc = [2,4,6,8]

moscow = (1, 2, 5, 7, 9)

print(type(nyc))
print(type(moscow))
```

```python
t1 = ()
t2 = ("hamza", 5, 4.5, True)
t3 = (1, "data", [1,2,3])
print(t1)
print(t2)
print(t3)
```

```python
# creating a tuple containing sinle elemnt

t4 = (32,)
print(type(t4))
```

```python
# type casting # changing the type of data or data structure

doha = [3,7,87,"abbas"]
tehran = tuple(doha)
print(type(doha))
print(type(tehran))
```

```python
# accessing elemets in tuples # indexing 

samra = (2, 4, 6, 8,10)
print(samra[2])

print(samra[4])


```

```python
# slicing in tuples

print(samra[0:2])
print(samra[:4])
print(samra[0:])
```

```python
# immutability\\
    
bombay = (8, 6, 4, 2,10, 10.5)
bombay[2]=6
```

```python
dhaka = (1, 2, 3, 4, 5, [7,7.5,.9])
dhaka[5].append(77)
print(dhaka)
```

```python
# operations in tuples
print(samra)

print(samra.count(8))

rio = (1,1,2,2,4,4,4)

print(rio.count(4))

print(rio.index(4))
```

```python
t1 = (1,2)
t2 = (3,4)

# concatenate

print(t1+t2)

# repetition

print(t2*3)

# membership

print(1 in rio)
```

```python
print(5 in rio)

# length

print(len(rio))

#min
print(min(rio))

#max

print(max(rio))
```

```python
# packing and unpacking of tuples

intro = ("hamza", 21, "reading", "the brothers karamazov")

# unpacking

name, age, hobby, current_read = intro

print(current_read)
```

```python
# iteration

lisbon = (22, 33, 44, 55, 66,"hamza","abdullah")

for i in lisbon:
    print(i)
    
```

```python
# nested_tuples

nested_list = [
    [1,2,3],
    [3,4,5]
]

print(nested_list)
```

```python
# nested tuple

nested_tuple = ((1,3,5),(2,4,6),(2,3,5))
print(nested_tuple)
```
