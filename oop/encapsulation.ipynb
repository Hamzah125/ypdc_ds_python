```python
# encapsulation

class Student:
    def __init__(self, name, grade):
        self.name = name # public
        self.__grade = grade # variable private
        # GET METHOD

    def get_grade(self):
        return self.__grade
# set 
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Invalid grade. Please enter a grade between 0 and 100.")
```

```python
S2 = Student("Hamza", "B")
print(S2.name) #open
print(S2.get_grade()) # via method
S2.set_grade(87) # valid update
S2.set_grade(105) # invalid update
```

```python
s1 = Student("Alice", "A")
print(s1.name) # Accessing public member
print(s1.get_grade) # Trying to access private member (will raise an AttributeError)
```

```python
# healthcare

# healthcare

class Patient:
    def __init__(self, name, age, disease):
        self.name = name
        self.age = age
        self.__disease = disease   # private variable

    @property
    def disease(self):
        # Publicly, the disease info is hidden
        return "Confidential"

    def allow(self, role):
        # Only Doctors can see the real disease
        if role == "Doctor":
            return self.__disease
        else:
            return "Access Denied"


# Usage
p1 = Patient("Ali", 30, "Diabetes")

print("Name:", p1.name)
print("Age:", p1.age)
print("Public Disease Info:", p1.disease)        # via @property
print("Doctor Access:", p1.allow("Doctor"))      # ✅ can see disease
print("Nurse Access:", p1.allow("Nurse"))        # ❌ denied

```

```python
# getter → to get the value of private variable
# setter → to update the value of private variable

class BankAccount:
    def __init__(self, holder, balance):
        self.holder = holder
        self.__balance = balance   # private variable

    # getter
    def get_balance(self):
        return self.__balance

    # setter
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Invalid balance amount")


# usage
acc = BankAccount("John Doe", 1000)

print("Initial Balance:", acc.get_balance())   #  via getter
acc.set_balance(2000)                          # via setter
print("Updated Balance:", acc.get_balance())

acc.set_balance(-500)                          
```
