```python
# finding average of crop yeild

crop_yeild = [100 , 600, 200, 600]
total= 0

for tons in crop_yeild:
    total += tons 
    average = total / len(crop_yeild)
    
print("average crop yeild is", average)
```

while loop is used to repeat a block of code as long as conditions fullfill or is true, take s iteration not in account

how does it work 

check the condition
run the specific block of code 
check the condition again
run till condition is true

```python
# case usage for agriculture, soil sciences

moisture = 45

while moisture <= 65:
    print("needs water")
    moisture += 5

print("moisture level ok",moisture )


```

```python
# genetic example for while loop

dna_sequence = "AGTCGTTGGCAAAAGCTTTGT"
index = 0

while index < len(dna_sequence):
    print(dna_sequence[index])
    index += 1
```

```python
# data science

temp = [24 , 19 , 18 ,38, 30 , 19, 34]
index = 0

while index < len(temp) and temp[index] <= 30:
    print("cool day", temp[index])
    index += 1
    
```

```python
# pass # stay here and do nothing # simply a placeholder

for i in range(19):
    pass
```

```python
# continue # skips the present iteration and do perform its function on next one

user_name = ["dimitri", "ivan", "karamazov", "", "alyosha", "", "gurshenka"]

for names in user_name:
    if names == "":
        continue
    print("welcome", names)
```

```python
# breaks # stops when reach a specific or specified condition

moisture_level = 30,45, 55, 65 , 70 , 85

for moisture in moisture_level:
    if moisture > 65:
        print("stop watering", moisture)
        break
    print("continue watering")
```
