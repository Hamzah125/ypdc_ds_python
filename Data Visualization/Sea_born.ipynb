```python
# scatter

sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="island", palette= "Set3")
sns.set_style("dark")
sns.set_context("notebook")
```

```python
# pallete set1 ;bright , set2 :soft, set 3, mix of bright and soft

# you will choose palette type based on your graphs 
# qualitative
# sequential
# diverging 
```

```python
sns.scatterplot(data=penguins, x="flipper_length_mm", y="bill_length_mm", hue="island", palette="Set2")
sns.set_style("dark")
```

```python
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="island", palette= "Set1", style="species")
```

```python
# stripplot

sns.stripplot(data = penguins, x= "species", y = "body_mass_g", hue = "island", dodge=True, jitter=True)
```

```python
# swarm plot

sns.set_style("whitegrid")
sns.set_context("talk")
sns.swarmplot(data=penguins, x="species", y="body_mass_g", hue="island", dodge=True)
sns.despine(left=True)
```

```python
# univariant

sns.histplot(data=penguins, x= "body_mass_g", hue="species", multiple="stack", kde=True)
```

```python
# kde also know as regression used to see the distribution of the data IN MORE EFFECIENT MANNER
sns.set_context("notebook")
sns.regplot(data = penguins, x = "bill_depth_mm", y = "body_mass_g", color= "green")
```

```python
# to see in a continuous manner \#lineplot

sns.set_style("whitegrid")
sns.lineplot(data=penguins, x="flipper_length_mm", y="body_mass_g", style="species", hue = "island")
```

```python
# joint plot

sns.jointplot(data= penguins, x = "body_mass_g", y="flipper_length_mm", hue="sex", kind="kde")
```

```python
# barplot # similar to histogram

sns.barplot(data=penguins, x="species", y="body_mass_g", hue="species", estimator=np.sum)
```

```python
# different species of penguins

#countplot

sns.countplot(data=penguins, x="island", hue="species", palette="Set2")
```

```python
# outliers detection # boxplot


sns.boxplot(data=penguins, x="species", y="body_mass_g", hue = "island")
```

```python
# violin plot # width shows the density

sns.violinplot(data= penguins, x="species", y="flipper_length_mm", split=True, hue="island")
```

```python
# swarm plot

sns.swarmplot(data=penguins, x="species", y="flipper_length_mm", hue="island", dodge=True)
```

```python
# kde for probability density function simply to see distributiom 

sns.kdeplot(data=penguins, x="flipper_length_mm")
sns.despine()
```

```python
# 2d  plots to show relationship between variables
# to show relationship between two nuerical columns


column = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]


```

```python
penguins[column].corr()
```

```python
# heatmap


sns.heatmap(penguins[column].corr(), annot=True, cmap="coolwarm", center=0, linewidths="2",vmin=-1, vmax=1)
```

```python
# regression 

sns.regplot(data=penguins, x="flipper_length_mm", y="body_mass_g")
```

```python
# rug plot

sns.rugplot(data=penguins, x="flipper_length_mm", height= 0.5)
```

```python
# pair plot

sns.pairplot(penguins, hue="species", diag_kind="kde", markers=["o", "s", "D"])
```

```python
# pair grid

# subplot of pair

graph = sns.PairGrid(data =penguins, hue="sex", palette="Set2")
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot)
graph.map_upper(sns.scatterplot)
```
