```python
# excessing elements in list

animals = ["lion", "tiger", "elephant", "hyenna","girrafe", "wolf"]
print(animals)

print(animals[-1])
```

```python
# slicing 

print(animals[0:3])
print(animals[2:6])
print(animals[0::2])
print(animals[2:])
print(animals[:5])
```

```python
# modify the elements
animals[4] = "crocodile"
#print(animals)

# append

animals.append("girrafe")
#print(animals)

# insert

animals.insert(2,"rhino")
#print(animals)

#pop

output = animals.pop()
print(output)


```

```python
# nested lists # lists within lists

list = [
    [1, 2, 3],
    [5,6,7],
    [11, 12, 13]
]
print(list[1][0])
```

```python
# iteration of a list

fruit = ["lychee", "grapes", "apple", "peach"]

for f in fruit:
    print(f.upper(), end = "")
```

```python
# list comprehension

# as a school teacher you wanna give grace marks

marks = [30, 40, 45 ,23, 41]
new_marks = []

for x in marks:
    new_marks.append(x+2)
    
print(new_marks)
                     
                    




```

```python
new_marks=[x+2 for x in marks]
print(new_marks)
```

```python
cube = []
for x in range (10):
    cube.append(x**3)

print(cube)
```

```python
easy =[x**3 for x in range(1,11)]
print(easy)
```

```python
# real world examples

# biology

bases = ["A", "T", "C", "G", "C", "T"]

for b in bases:
    print(b)
```

```python
# expenses

expense = [3000 , 5000, 4500, 5300]

total= sum(expense)
print("my total monthly expense", total)
```

```python
# list comprehension

# converting kgs into tons

crop_yeild_tons = [1200 , 5400, 6700, 6900]
crop_yeild_kgs = [y*1000 for y in crop_yeild_tons]
print(crop_yeild_kgs)
```

```python
dna_bases = ["A","C","T","G"]
purines = [b for b in dna_bases if b in ["A","G"]]
print(purines)
```

```python
# nested LISTSere we have loops inside loops

# agriculture

yeild= [
    [123, 300, 500],
    [200, 400, 600],
    [100, 300, 700]
]
all_yeild=[value for field in yeild for value in field]
print(sum(all_yeild))
    
```
