```python
# for loop in agriculture

yields = [2600, 4300, 2800, 3500]  # kg per hectare
total = 0

for crop in yields:
    total += crop

average_yield = total / len(yields)
print("Average Crop Yield:", average_yield, "kg/ha")
```

```python
# while loops
# A while loop is used to repeat a block of code as 
# long as a given condition is True

# while condition

# how it works
#check condition — If it’s True, run the code block.

#Run code block.

#Check condition again.

#Repeat until the condition becomes False.
```

```python
#example
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
#keep watering untill moisture level reaches 60
moisture = 45

while moisture <= 60:
    print("Watering crops...")
    moisture += 5
print("Moisture level OK", moisture)
```

```python
# biological # count till all are read

dna = ["A", "T", "C", "G", "A", "T"]
index = 0
while index < len(dna):
    print(dna[index])
    index += 1
```

```python
#Read temperature data until a hot day is found
temps = [25, 27, 29, 33, 28]
index = 0
while index < temps[index] <= 30:
    print("Cool day:", temps[index])
    index += 1
```

```python
dna = ["A", "T", "G", "C", "A", "T"]  
index = 0

while index < len(dna):       # Keep looping until we've read all bases
    print(dna[index])         # Show the current base
    index += 1                # Move to the next base
```

Loop control statements change the normal flow of loops.
They are used to:

Stop the loop early (break)

Skip one iteration (continue)

Do nothing temporarily (pass)

```python
#Immediately stops the loop, even if the loop condition is still True.
moisture_levels = [50, 55, 65, 80, 90]

for level in moisture_levels:
    if level > 75:
        print("Alert! Moisture too high:", level)
        break
    print("Moisture OK:", level)
```
