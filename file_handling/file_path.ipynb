```python
# CSV

import csv
```

```python
with open("crops.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["crop", "yeild"])
    writer.writerow(["wheat", 200])
    writer.writerow(["corn", 300])
    writer.writerow(["soybean", 150])
```

```python
with open('crops.csv',"r") as file:
    reader= csv.reader(file)
    for row in reader:
        print(row)
```
