```python
# constructor #__init__ # a special method that runs automatically when an object
# is created #actually used to initialize the attribues 

class Book:
    def __init__(self, title, author, genre):
        self.title = title
        self.author = author
        self.genre = genre

    def display_info(self):
        print(f"Book: {self.title} by {self.author} of Genre: {self.genre}")
```

```python
# object 
book1 = Book("1984", "George Orwell", "Dystopian Novel")
book2 = Book("Catch 22", "Joseph Heller", "Satirical Novel")
book1.display_info()
book2.display_info()
    
```

```python
# class and object

class Car:
    def __init__(self, make, model, variant):
        self.make = make
        self.model = model
        self.variant = variant
    def drive(self):
        print(f"Driving a {self.make} {self.model} {self.variant}")
```

```python
# object 

car1 = Car("toyota", "lc", "zx")

print(car1.make)
print(car1.model)
print(car1.variant)

car1.drive()

dir(car1)
```

```python
# bank account

class BankAccount:
    def __init__(self, account_holder, account_number, account_type, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.account_type = account_type
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance is {self.balance}.")
    def check_balance(self):
        print(f"Current balance is {self.balance}.")
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
        else:
            print(f"Withdrawing {amount}.")
            self.balance -= amount
            print(f"current balance is {self.balance}.")
```

```python
# object

acc1 = BankAccount("hamza", "4321", "current", 1000000000)
```

```python
print(acc1.account_holder)
print(acc1.account_number)
print(acc1.account_type)
print(acc1.balance)
```

```python
acc1.deposit(5000)
```

```python
acc1.withdraw(1000000)
acc1.check_balance()
```

```python
# student class and object

class Student:
    def __init__(self, name, age, degree):
        self.name = name
        self.age = age
        self.degree = degree
    def study(self):
        print(f"{self.name} is in {self.degree} department.")
```

```python
# objects

s1 = Student("hamza", 21, "law")
s2 = Student("ali", 22, "computer science")

print(s1.name, s1.age, s1.degree)

s1.study()
```

```python
# biological example of class and object

class DNA:
    def __init__(self,sequence):
        self.sequence = sequence
    def get_length(self):
        return len(self.sequence)
```

```python
# object

dna1 = DNA("ATGCCGTACGAT")

print(dna1.get_length())
```
