```python
# try and except block

try:
    z = x
except NameError:
    print("Variable x is not defined")
```

```python
# zero division error
try:
    result = 10 / 0
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
# file presence error

try:
    with open('example.txt', "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Error: File not found.")
```
