```python
# single inheritence # in which one child class inherits a parent

class Animal:
    def eat(self):
        print("This animal eats food.")


```

```python
# dog inherit animal

class Dog(Animal):
    def bark(self):
        print("The dog barks.")
```

```python
d = Dog()
d.eat()
d.bark()
```

```python
# another example

class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model 
        self.year = year
    def display_info(self):
        print(f"Car Information: {self.year} {self.make} {self.model}")
```

```python
class SUV(Car):
    def __init__(self, make, model, color):
        super().__init__(self, make, model)
        self.color = color
    def start(self):
        print(f"The {self.model} SUV is starting.")
```

```python
t1 = SUV("toyota", "surf", "white")
print(t1.color)
t1.start()
print(t1.make)
```

```python
# multiple inheritence 

class Animal:
    def __init__(self,name):
        self.name = name
    def voice(self):
        print(f"{self.name} makes a sound.")
```

```python
class Pet:
    def __init__(self, owner):
        self.owner = owner
```

```python
# daughter class

class Cat(Animal, Pet):   # multiple inheritance
    def __init__(self, name, owner):
        Animal.__init__(self, name)  # initialize Animal
        Pet.__init__(self, owner)    # initialize Pet

d1 = Cat("whisker", "bob")
d2 = Cat("fluffy", "alice")

print(d1.name, d1.owner)  # whisker bob
print(d2.name, d2.owner) 
```

```python
class Father:
    def skill(self):
        print("Father:Farming")
        
class Mother:
    def skill(self):
        print("Mother:Cooking")
        
class Child(Mother, Father):
    pass

c = Child()
c.skill()  # MRO
```

```python
# Grandparent class
class University:
    def __init__(self, uni_name):
        self.uni_name = uni_name

    def show_uni(self):
        print(f"University: {self.uni_name}")

# Parent class
class Department(University):
    def __init__(self, uni_name, dept_name):
        super().__init__(uni_name)
        self.dept_name = dept_name

    def show_dept(self):
        print(f"Department: {self.dept_name}")

# Child class
class Student(Department):
    def __init__(self, uni_name, dept_name, student_name):
        super().__init__(uni_name, dept_name)
        self.student_name = student_name

    def show_student(self):
        self.show_uni()
        self.show_dept()
        print(f"Student: {self.student_name}")

# Testing
s1 = Student("UAF", "Microbiology", "Hamza")
s1.show_student()
```
